## 8. Does it have missing values? If yes, which column, which row.


```python
missing = df.isnull().sum()
print("Missing values in each column:")
print(missing)
print("\nColumns with missing values:")
print(missing[missing > 0])
print("\nNumber of columns with missing values:", (missing>0).sum())
```

```python
#axis=1 means checking the rows horizontally one by one
#axis=0 means going down the columns
rows_missing = df[df.isnull().any(axis=1)]
print("Number of rows with at least one missing value:", len(rows_missing))
print("First 10 rows containing missing values:")
display(rows_missing.head(10))

# To find exact positions for one particular column:
column_to_check = "Age"
print("Passenger IDs with missing", column_to_check,":")
print(df.loc[df[column_to_check].isnull(), "PassengerId"].tolist())
```

Missing values occur in 3 columns,which is Age(177), Cabin(687),Embarked(2) and the actual rows = Passengers Id -1

## 9. What kind of question can be asked, should we ask. Does the dataset suits the question we are asking?

How did the survival rates differ from the sex?

```python
import matplotlib.pyplot as plt

survival_rate = df.groupby("Sex")["Survived"].mean()*100
print("Percentage who survived in each class:")
print(survival_rate.round(1))

survival_rate.plot(kind="bar")
plt.title("Survival percentage by passenger sex")
plt.xlabel("Passenger Sex")
plt.ylabel("Survival Percentage(%)")
plt.xticks(rotation=0)
plt.show()
```

Yes, the dataset suits this question because the dataset contains the target variable "Survived" (0=perished, 1=survived) and unlike age or Cabin, the "Sex" column has zero missing value in order to provide a complete statistical integrity
